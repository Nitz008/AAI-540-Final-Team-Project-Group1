# EMBER2024: exploratory data analysis

**Data provenance notice:** This notebook analyzes a reconstructed approximation derived from the available processed CSV files. The original raw JSONL files were unavailable for this run. It does not claim to reproduce raw-data validation or the original feature-preparation pipeline.

The analysis uses the versioned files under `data/processed/ember2024_reconstructed_v1`. Interpret the plots as exploratory analysis of these reconstructed features, not as proof of raw-data quality or production model performance.

```python
from pathlib import Path
import json
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

PROJECT_ROOT = next(
    (p for p in (Path.cwd(), *Path.cwd().parents)
     if (p / "data" / "processed" / "ember2024_reconstructed_v1").is_dir()),
    None,
)
if PROJECT_ROOT is None:
    raise FileNotFoundError(
        "Could not locate data/processed/ember2024_reconstructed_v1. "
        "Open this notebook from the project directory."
    )

DATA_DIR = PROJECT_ROOT / "data" / "processed" / "ember2024_reconstructed_v1"
REPORT_PATH = DATA_DIR / "reconstruction_report.json"
SCHEMA_PATH = DATA_DIR / "feature_schema_reconstructed.json"

with REPORT_PATH.open(encoding="utf-8") as f:
    reconstruction_report = json.load(f)
with SCHEMA_PATH.open(encoding="utf-8") as f:
    feature_schema = json.load(f)

splits = {
    name: pd.read_csv(DATA_DIR / f"{name}.csv")
    for name in ("train", "validation", "test", "reserved_production")
}
train = splits["train"]
validation = splits["validation"]
test = splits["test"]
reserved = splits["reserved_production"]
feature_columns = feature_schema["model_feature_columns"]

sns.set_theme(style="whitegrid", context="notebook")

print("Project root:", PROJECT_ROOT)
print("Data directory:", DATA_DIR)
print("Reconstruction status:", reconstruction_report.get("status", "Not specified"))
print("Feature count:", len(feature_columns))
print("Rows per split:", {name: len(frame) for name, frame in splits.items()})
print("Important: reconstructed approximation, not regenerated raw data.")
pd.DataFrame([
    {"split": name, "rows": len(frame),
     "benign_label_0": int((frame["label"] == 0).sum()),
     "malicious_label_1": int((frame["label"] == 1).sum()),
     "week_start_min": frame["source_week_start"].min(),
     "week_start_max": frame["source_week_start"].max()}
    for name, frame in splits.items()
])
```

## Dataset and class distribution

Counts below describe the available reconstructed CSV splits only. They are not raw-dataset validation counts.


```python
class_counts = (
    train["label"].value_counts().sort_index()
    .rename(index={0: "Benign (0)", 1: "Malicious (1)"})
    .rename("records").rename_axis("class").reset_index()
)
display(class_counts)
ax = sns.barplot(data=class_counts, x="class", y="records", hue="class", legend=False)
ax.set(title="Reconstructed training split: class distribution",
       xlabel="Class", ylabel="Records")
plt.tight_layout()
plt.show()

display(pd.DataFrame([
    {"split": name, "rows": len(frame),
     "class_0": int((frame["label"] == 0).sum()),
     "class_1": int((frame["label"] == 1).sum())}
    for name, frame in splits.items()
]))
```

## Missingness and schema

The reconstructed CSVs contain model features after preprocessing. Missing values may already have been imputed, so this section reports missing values that remain in the reconstructed files; it cannot recover the original raw-field missingness.


```python
missing_summary = pd.DataFrame({
    "missing_count": train[feature_columns].isna().sum(),
    "missing_rate": train[feature_columns].isna().mean(),
    "unique_values": train[feature_columns].nunique(dropna=True),
}).sort_values("missing_rate", ascending=False)
display(missing_summary)

if missing_summary["missing_count"].sum():
    ax = sns.barplot(x=missing_summary["missing_rate"],
                     y=missing_summary.index, color="steelblue")
    ax.set(title="Remaining missing values in reconstructed training features",
           xlabel="Missing rate", ylabel="Feature")
    plt.tight_layout()
    plt.show()
else:
    print("No missing values remain in the reconstructed model-feature columns.")
display(train[feature_columns].describe().T)
```

## Feature distributions

These plots describe the reconstructed, standardized feature values. They should not be interpreted as original-scale PE measurements.


```python
display_columns = feature_columns[:8]
sample = train.groupby("label", group_keys=False).sample(
    n=min(5000, int(train["label"].value_counts().min())), random_state=540
)
fig, axes = plt.subplots(len(display_columns), 2,
                         figsize=(14, max(4, 2.8 * len(display_columns))))
for row, column in enumerate(display_columns):
    sns.histplot(data=sample, x=column, hue="label", stat="density",
                 common_norm=False, element="step", ax=axes[row, 0])
    axes[row, 0].set_title(f"{column}: distribution by class")
    sns.boxplot(data=sample, x="label", y=column, ax=axes[row, 1])
    axes[row, 1].set_title(f"{column}: class comparison")
plt.tight_layout()
plt.show()
display(train[feature_columns].describe().T)
```

## Correlation and redundancy

Correlation is calculated on a deterministic sample of reconstructed training rows. It is a descriptive review signal, not an automatic feature-exclusion rule.


```python
corr_sample = train[feature_columns].sample(
    n=min(20000, len(train)), random_state=540
)
correlation = corr_sample.corr(numeric_only=True)
plt.figure(figsize=(12, 9))
sns.heatmap(correlation, cmap="vlag", center=0, vmin=-1, vmax=1)
plt.title("Reconstructed training-feature correlations")
plt.tight_layout()
plt.show()

upper = correlation.where(
    pd.DataFrame(
        [[i < j for j in range(len(correlation.columns))]
         for i in range(len(correlation.columns))],
        index=correlation.index, columns=correlation.columns
    )
)
high_corr = (
    upper.stack().rename("correlation").reset_index()
    .rename(columns={"level_0": "feature_1", "level_1": "feature_2"})
)
high_corr["absolute_correlation"] = high_corr["correlation"].abs()
display(high_corr.sort_values("absolute_correlation", ascending=False).head(25))
```

## Temporal coverage and leakage review

The chart summarizes week coverage recorded in the reconstructed CSVs. It does not independently validate original timestamps or the raw EMBER2024 partition. Exclude identifiers and any feature unavailable at inference time.


```python
temporal_rows = []
for name, frame in splits.items():
    dates = pd.to_datetime(frame["source_week_start"], errors="coerce")
    for label, count in frame["label"].value_counts().items():
        temporal_rows.append({
            "split": name, "label": int(label), "records": int(count),
            "week_start_min": dates.min(), "week_start_max": dates.max()
        })
temporal_summary = pd.DataFrame(temporal_rows)
display(temporal_summary)

train_timeline = train.copy()
train_timeline["source_week_start"] = pd.to_datetime(
    train_timeline["source_week_start"], errors="coerce"
)
train_timeline["month"] = train_timeline["source_week_start"].dt.to_period("M").astype(str)
monthly = train_timeline.groupby(["month", "label"]).size().reset_index(name="records")
plt.figure(figsize=(14, 5))
sns.lineplot(data=monthly, x="month", y="records", hue="label", marker="o")
plt.xticks(rotation=45, ha="right")
plt.title("Reconstructed training records by source month and class")
plt.tight_layout()
plt.show()
```
